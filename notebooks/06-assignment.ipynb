## 執行測試！

````python
from datetime import datetime

class TestChapter06(unittest.TestCase):
    def test_0601(self):
        self.assertTrue(answer_0601)
    def test_0602(self):
        self.assertEqual(answer_0602, 4)
    def test_0603(self):
        df = pd.read_sql(answer_0603, conn)
        self.assertEqual(df.shape[1], 4)
        m = df.shape[0]
        for i in range(m):
            self.assertIn("-KY", df.iloc[:, 2].values[i])
    def test_0604(self):
        df = pd.read_sql(answer_0604, conn)
        m, n = df.shape
        self.assertEqual(n, 3)
        for i in range(m):
            self.assertIn("00", df.iloc[:, 1].values[i])
    def test_0605(self):
        df = pd.read_sql(answer_0605, conn)
        m, n = df.shape
        self.assertEqual(n, 2)
        for i in range(m):
            self.assertIn(".TWO", df.iloc[:, 0].values[i])
            self.assertIn("-KY", df.iloc[:, 1].values[i])
    def test_0606(self):
        df = pd.read_sql(answer_0606, conn)
        m, n = df.shape
        self.assertEqual(n, 3)
        for i in range(m):
            condition_1 = df.iloc[:, 1].values[i] > 1000000000000
            condition_2 = df.iloc[:, 2].values[i] > 1000
            self.assertTrue(condition_1 or condition_2)
    def test_0607(self):
        df = pd.read_sql(answer_0607, conn)
        m, n = df.shape
        self.assertEqual(n, 3)
        lower_date_obj = datetime.strptime("2000-01-01", "%Y-%m-%d")
        lower_date_integer = int(lower_date_obj.strftime("%Y%m%d"))
        upper_date_obj = datetime.strptime("2019-12-31", "%Y-%m-%d")
        upper_date_integer = int(upper_date_obj.strftime("%Y%m%d"))
        for i in range(m):
            date_obj = datetime.strptime(df.iloc[:, 2].values[i], "%Y-%m-%d")
            date_integer = int(date_obj.strftime("%Y%m%d"))
            condition_1 = date_integer < lower_date_integer
            condition_2 = date_integer > upper_date_integer
            self.assertTrue(condition_1 or condition_2)
suite = unittest.TestLoader().loadTestsFromTestCase(TestChapter06)
runner = unittest.TextTestRunner(verbosity=2)
test_results = runner.run(suite)
number_of_failures = len(test_results.failures)
number_of_errors = len(test_results.errors)
number_of_test_runs = test_results.testsRun
number_of_successes = number_of_test_runs - (number_of_failures + number_of_errors)
chapter_index = 6
chapter_name = "篩選資料"
````
